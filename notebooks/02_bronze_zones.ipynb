```sql
CREATE OR REPLACE TABLE workspace.bronze.zones AS
SELECT
    *
  , _metadata.file_path   AS _source_file
  , current_timestamp()   AS _ingested_at
FROM
    read_files(
        '/Volumes/workspace/default/raw/zones/taxi_zone_lookup.csv'
      , format => 'csv'
      , header => true
    );
```

```sql
SELECT
    *
FROM
    workspace.bronze.zones
ORDER BY
    LocationID;
```
